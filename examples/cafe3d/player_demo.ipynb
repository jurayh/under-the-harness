**Embedding this in a chapter notebook** takes the same cell:
point the iframe at `player.html?filmstrip=careful` (or `gist`),
keep the poster `<img>` under it, and the cell degrades to the
still scene wherever scripts don't run. Paths are relative to
this folder; a chapter notebook elsewhere in the repo needs the
`examples/cafe3d/` prefix — or its own copy of the folder.

What a chapter 7 integration still needs is listed in
`CAFE3D_PLAYER_REPORT.md`: the filmstrips here are exported
